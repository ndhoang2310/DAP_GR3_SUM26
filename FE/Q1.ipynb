```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

df = pd.DataFrame({
    "Color": ["Red", "Blue", "Green", "Red", "Blue", "Green", "Red"]
})

# A
df["A"] = df["Color"].map({
    "Red": 0,
    "Blue": 1,
    "Green": 2
})

# B
le = LabelEncoder()
df["B"] = le.fit_transform(df["Color"])

# C
df["C"] = df["Color"].apply(
    lambda x: {
        "Red": 2,
        "Blue": 0,
        "Green": 1
    }[x]
)

# D
df["D"] = df["Color"].astype("category").cat.codes

print(df)

print("A:", df["A"].unique())
print("B:", df["B"].unique())
print("C:", df["C"].unique())
print("D:", df["D"].unique())

print("LabelEncoder classes:", le.classes_)
```
